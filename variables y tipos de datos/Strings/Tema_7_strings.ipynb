Polimorfismo
============
El signo más (+) tiene significados distintos según el objeto: suma en el caso de los números y concatenación en el de las cadenas.

Esta es una propiedad general de Python denominada polimorfismo.

El significado de una operación depende de los objetos sobre los que se realiza.

Como verás cuando estudiemos el tipado dinámico, esta propiedad del polimorfismo explica en gran medida la concisión y la flexibilidad del código de Python.

Dado que los tipos no están restringidos, una operación escrita en Python normalmente puede funcionar automáticamente con muchos tipos diferentes de objetos, siempre que estos admitan una interfaz compatible (como la operación + en este caso).

Inmutabilidad
=============
Los objetos inmutables no se pueden modificar.

En Python, todos los objetos se clasifican como inmutables (inalterables) o no.

En cuanto a los tipos básicos:

los números, las cadenas y las tuplas son inmutables;
las listas, los diccionarios y los conjuntos son mutables.

Por ejemplo ahora si intentamos cambiar el valor de uno de los caracteres del strin s nos saltara un IndexError como en el siguiente caso :

```python
s[0] = 'z'
```

Pero podemos usar expresiones para crear nuevos objetos 

```python
s = 'Z' + s[1:]
s
```

En este caso lo que hacemos es crear un nuebo objeto donde al caracter Z le concatenamos todos los caracteres entre la posicion [1:] por lo que s deja de apuntar al valor en memoria SPAM para apuntar a ZPAM
